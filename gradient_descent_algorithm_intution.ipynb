```python
m = 78.35
b=0

loss_slope = -2 * np.sum(y - (78.35 * X.ravel()) - b)
loss_slope
```

#### Learning rate - 0.1

```python
lr_rate = 0.1
step_size = loss_slope * lr_rate
step_size
```

```python
b_new = b - step_size
b_new
```

### b = 20.92

```python
y_pred_b_20 = ((78.35 * X) + b_new).reshape(4,)

plt.scatter(X,y)
plt.plot(X,lr.predict(X),color='red',label='OLS')
plt.plot(X,y_pred_b_0,color='#00a65a',label='b = 0')
plt.plot(X,y_pred_b_20,color='#ff32aa',label='b = 20')
plt.legend()
plt.show()
                                        
```

## Iteration 2

```python
loss_slope = -2 * np.sum(y - (78.35 * X.ravel()) - b_new)
loss_slope
```

```python
step_size = loss_slope * lr_rate
step_size 
```

```python
b_new_20 = b_new - step_size
b_new_20
```

## b = 25

```python
y_pred_b_25 = ((78.35 * X) + b_new_20).reshape(4,)

plt.scatter(X,y)
plt.plot(X,lr.predict(X),color='red',label='OLS')
plt.plot(X,y_pred_b_0,color='#00a65a',label='b = 0')
plt.plot(X,y_pred_b_20,color='#ff32aa',label='b = 20')
plt.plot(X,y_pred_b_25,color='#00ff31',label='b = 25')
plt.legend()
plt.show()
```

## Iteration 3

```python

b = -100
m = 78.35
lr = 0.01

epochs = 100

for i in range(epochs):
  loss_slope = -2 * np.sum(y - m*X.ravel() - b)
  b = b - (lr * loss_slope)

  y_pred = m * X + b

  plt.plot(X,y_pred,label=b)
plt.scatter(X,y)
```
